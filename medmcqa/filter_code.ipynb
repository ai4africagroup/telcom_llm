```python
from langchain_community.llms import Ollama
import json
llm = Ollama(model = "llama3")
```

```python
# llm.invoke("Tell me 3 red flower names.")
with open("train.json", "r") as f:
    test_data = f.readlines()
print(len(test_data))
```

```python
cleaned = []
prompt = "In the follwing Text, if it starts with an option ID, remove it and return the text as it is. Only output the cleaned text without any comments. Text: "
for line in test_data[:5000]:
    sample = json.loads(line)
    if sample["exp"] != None and (sample["exp"].lower().startswith("ans") or sample["exp"].lower().startswith("option")):
        text = sample["exp"]
        print(text)
        if sample["exp"].startswith("Ans. is"):
            cleaned_text = (' '.join(text.split(" ")[3:]))
        else:
            cleaned_text = (' '.join(text.split(" ")[2:]))
        print(cleaned_text)
        print("="*100) 
        sample["exp"]  = cleaned_text
        cleaned.append(sample)
    else:
        cleaned.append(sample)

        
        
```

```python
# Write the cleaned list to a file in the required format
with open("train5000_cleaned.json", "w") as file:
    for item in cleaned:
        json.dump(item, file)
        file.write("\n")
```
